```python
import folium
from folium import plugins
```

```python
sete_maravilhas_antigas = [
    "Grande Pirâmide de Gize",
    "Jardins Suspensos da Babilônia",
    "Templo de Ártemis",
    "Estátua de Zeus em Olímpia",
    "Mausoléu de Halicarnasso",
    "Colosso de Rodes",
    "Farol de Alexandria"
]

sete_maravilhas_coordenadas = [
    [29.9792, 31.1348],
    [31.532, 44.455],
    [37.9498, 27.3639],
    [37.638, 21.6574],
    [37.038, 27.4515],
    [36.4512, 28.2552],
    [31.2139, 29.9129]
]
```

```python
mapa = folium.Map(location=[33.7208, 28.9086], zoom_start=5, control_scale=True)

# Adicionando camada de tiles personalizada
folium.TileLayer(
    tiles='https://server.arcgisonline.com/ArcGIS/rest/services/World_Street_Map/MapServer/tile/{z}/{y}/{x}',
    attr='Tiles © Esri — Source: Esri, DeLorme, NAVTEQ, USGS, Intermap, iPC, NRCAN, Esri Japan, METI, Esri China (Hong Kong), Esri (Thailand), TomTom, 2012',
    name="Esri.WorldStreetMap"
).add_to(mapa)

# Marcadores com popup correto
for i in range(len(sete_maravilhas_coordenadas)):
    coordenada = sete_maravilhas_coordenadas[i]
    maravilha = sete_maravilhas_antigas[i]
    folium.Marker(coordenada, popup=maravilha, tooltip="Clique para ver").add_to(mapa)

# Ativar popup de coordenadas ao clicar
mapa.add_child(folium.LatLngPopup())

# Salvar o mapa
mapa.save("./mapas/maravilhas_antigas.html")

mapa
```
