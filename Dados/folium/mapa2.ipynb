```python
import folium
from folium import plugins
```

```python
sete_maravilhas_novas = [
    "Coliseu",
    "A Grande Muralha da China",
    "Cristo Redentor",
    "Chichén Itzá",
    "Machu Picchu",
    "Petra ",
    "Taj Mahal"
]

novas_maravilhas_coordenadas = [
    [41.891, 12.4915],
    [40.4319, 116.5704],
    [-22.9519, -43.2105],
    [20.6843, -88.5678],
    [-13.1631, -72.5450],
    [30.3285, 35.4444],
    [27.1751, 78.0421]
]
```

```python
mapa2 = folium.Map(location=[-4.0102, -17.9297], zoom_start=2, control_scale=True)

# Adicionando camada de tiles personalizada
folium.TileLayer(
    tiles='https://server.arcgisonline.com/ArcGIS/rest/services/World_Street_Map/MapServer/tile/{z}/{y}/{x}',
    attr='Tiles © Esri — Source: Esri, DeLorme, NAVTEQ, USGS, Intermap, iPC, NRCAN, Esri Japan, METI, Esri China (Hong Kong), Esri (Thailand), TomTom, 2012',
    name="Esri.WorldStreetMap"
).add_to(mapa2)

# Marcadores com popup correto
for i in range(len(novas_maravilhas_coordenadas)):
    coordenada = novas_maravilhas_coordenadas[i]
    maravilha = sete_maravilhas_novas[i]
    folium.Marker(
        coordenada,
        popup=maravilha,
        tooltip="Clique para ver",
        icon=folium.Icon(color="pink", icon="star")
    ).add_to(mapa2)

# Ativar popup de coordenadas ao clicar
mapa2.add_child(folium.LatLngPopup())

# Salvar o mapa
mapa2.save("./mapas/maravilhas_novas.html")

mapa2
```
