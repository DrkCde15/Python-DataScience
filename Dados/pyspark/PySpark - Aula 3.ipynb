#Aula 3 - PySpark

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, desc, asc
```

```python
sprk = SparkSession.builder.appName("Spark DataFrames").getOrCreate # cria uma sessão Spark
```

```python
spark = SparkSession.builder.getOrCreate()
data = [("Julio", 20), ("Raphael", 30), ("Steph", 26)]
cols = ["Nome", "Idade"]
df = spark.createDataFrame(data = data, schema = cols) # cria um dataframe usando uma lista de tuplas
df.display() # exibe o dataframe
```

```python
df = df.withColumn("Nova Idade", col("Idade") + 5)
df.display()
```

```python
#transformação
df.select("Nome", "Idade").filter("Idade > 25").display()
```

```python
data_frame = [
    ("Raphael", "Senior DE", 10000),
    ("Steph", "Junior DE", 4700),
    ("Julio", "Estágiario DE", 2000),
    ("May", "Marketing", 7000),
    ("Clara", "Marketing", 5000)
]
cols = ["Nome", "Cargo", "Salario"]

df_new = spark.createDataFrame(data = data_frame, schema = cols)
df_new.display()
```

```python
#Filtrar funcionários com cargo
df_new.filter(df_new["Cargo"] == "Marketing").display()
```

```python
#Filtrar funcionários com salario

df_new.filter(df_new["Salario"] > 5000).display() 
```

```python
#Media dos salarios pelos nomes

df_new.groupBy("Nome").avg("Salario").display()
```

```python
#Media dos salarios pelo cargo

df_new.groupBy("Cargo").avg("Salario").display()
```
