###Aula 1 - PySpark: Carregamento

```python
# Leitura de arquivos CSV usando spark.read.format()
# Cada DataFrame recebe o caminho completo para o arquivo no Databricks
df1 = spark.read.format('csv').load('/Workspace/Users/jcesarsantana215@gmail.com/datasets/marca_carro.csv')
df2 = spark.read.format('csv').load('/Workspace/Users/jcesarsantana215@gmail.com/datasets/marcas_duplicadas.csv')
df3 = spark.read.format('csv').load('/Workspace/Users/jcesarsantana215@gmail.com/datasets/modelo_carro.csv')

# Exibe o DataFrame no formato de tabela (função display do Databricks)
display(df1)
```

```python
# Escrita de dados em CSV com modo 'overwrite'
# O modo 'overwrite' substitui completamente os dados existentes no caminho de destino
# Útil quando se deseja atualizar totalmente o conteúdo de uma tabela
df1.write.format('csv').mode('overwrite').save('/caminho/para/salvar')
df2.write.format('csv').mode('overwrite').save('/caminho/para/salvar')
df3.write.format('csv').mode('overwrite').save('/caminho/para/salvar')
```

```python
# Escrita de dados em CSV com modo 'append'
# O modo 'append' adiciona os novos dados aos já existentes no destino
# Cuidado: pode gerar duplicatas se os mesmos dados forem adicionados novamente
df1.write.format('csv').mode('append').save('/caminho/para/salvar')
df2.write.format('csv').mode('append').save('/caminho/para/salvar')
df3.write.format('csv').mode('append').save('/caminho/para/salvar')
```

```python
# Contagem de registros do DataFrame
# O método count() retorna o número total de linhas do DataFrame
print(df1.count())
```
