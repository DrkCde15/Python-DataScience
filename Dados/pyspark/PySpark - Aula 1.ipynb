#AULA 1 - PySpark

```python
data = [("Julio", 20), ("Steph", 26), ("Raphael", 30)]
cols = ["Nome", "Idade"]
df = spark.createDataFrame(data = data, schema = cols) # cria um dataframe usando uma lista de tuplas
df.display() # exibe o dataframe
```

```python
df.createOrReplaceTempView("tabela") # cria uma tabela temporária usando o dataframe
```

```python
%sql
SELECT * FROM tabela; --Ler a tabela
```
