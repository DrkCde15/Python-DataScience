###Dominando dados nulos

```python
from pyspark.sql.functions import *
```

```python
data = [("Julio", None, None),
        ("May", "Marketing", 7000),
        ("Clara", "Eventos", 3000),
        ("Stephanie", None, None),
        ("Raphael", "Engenheiro de Dados", 10000),
        ("Juliana", "Cabeleireira", 2000),
        ("Eduardo", None, None),
        ("Caio", None, None)]
columns = ["Nome", "Cargo", "Salario"]
df = spark.createDataFrame(data, schema=columns)
df.display()
```

```python
df = df.withColumn("Cargo eh nulo?", isnull(col("Cargo")))
df.display()
```

```python
df = df.withColumn('Salario eh nulo?', isnull(col('Salario')))

df.display()
```

```python
df = df.filter(col('Salario eh nulo?') == 'true').display()
```

```python
df = df.fillna(0, subset=['Salario'])
df.display()
```
