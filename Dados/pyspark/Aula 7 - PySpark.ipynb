###Aula 7 - PySpark: Like e Between

```python
# Setup completo: imports, leitura, normalização e tipagem
#Imports
from pyspark.sql.functions import *
from pyspark.sql.types import *

#Leitura de arquivos
df_carros = spark.read.format('csv').option('header', 'true').option('encoding', 'utf-8').load('/Workspace/Users/jcesarsantana215@gmail.com/datasets/modelo_carro.csv', sep =',')

#Normalizacao
df_carros = df_carros.withColumn(
    "preco",
    regexp_replace("preco", r"\$", "")
)

#Tipagem de dados
df_carros = df_carros.select(
    col('id_carro').cast(IntegerType()),
    col("preco").cast(DoubleType()),
    col("cod_marca").cast(IntegerType()),
    "modelo_carro")

display(df_carros)
```

```python
# Cria view para consultas SQL
df_carros.createOrReplaceTempView('carros')
```

```python
# Filtro LIKE: modelos que começam com 'E'
df_carros_sql_like = spark.sql(
    """
    SELECT * FROM carros WHERE modelo_carro LIKE 'E%'
    """)
display(df_carros_sql_like)
```

```python
# LIKE: modelos que contêm 'alo'
%sql
SELECT * FROM carros WHERE modelo_carro LIKE '%alo%'
```

```python
# LIKE: modelos que terminam com 'rt'
%sql
SELECT * FROM carros WHERE modelo_carro LIKE '%rt'
```

```python
# LIKE: modelos que começam com 'E'
%sql
SELECT * FROM carros WHERE modelo_carro LIKE 'E%'
```

```python
# Filtro BETWEEN: preço entre 50000 e 65000
%sql
SELECT * FROM carros WHERE preco BETWEEN 50000 AND 65000
```

```python
# BETWEEN usando PySpark SQL
df_carros_sql_btw = spark.sql(
    """
    SELECT * FROM carros WHERE preco BETWEEN 50000 AND 65000
    """)
display(df_carros_sql_btw)
```

```python
# Combinação de LIKE e BETWEEN no SQL
df_carros_sql = spark.sql(
    """
    SELECT * FROM carros WHERE modelo_carro LIKE 'E%'
    AND preco BETWEEN 50000 AND 65000
    """
)
display(df_carros_sql)
```

```python
# Filtro LIKE usando API PySpark
df_carros_spark = df_carros
df_carros_spark = df_carros_spark.where(
    col('modelo_carro').like('E%')
)
display(df_carros_spark)
```

```python
# Filtro BETWEEN usando API PySpark
df_carros_spark = df_carros
df_carros_spark = df_carros_spark.where(
    col('preco').between(50000, 65000)
)
display(df_carros_spark)
```
