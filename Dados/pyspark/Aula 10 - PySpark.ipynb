###Aula 10 - PySpark:Existis e Left Semi

```python
# Preparação dos DataFrames para demonstrar EXISTS e LEFT SEMI
# Imports
from pyspark.sql.functions import *

# Leitura de arquivos
df_carros = (
    spark.read
    .format('csv')
    .option('header', 'true')
    .option('encoding', 'utf-8')
    .load('/Workspace/Users/jcesarsantana215@gmail.com/datasets/modelo_carro.csv', sep=',')
    .distinct()
)

df_carros_source = df_carros.where(
    (col("id_carro") == '1') |
    (col("id_carro") == '2') |
    (col("id_carro") == '3')
)

df_carros_final = df_carros.where(
    (col("id_carro") == '1') |
    (col("id_carro") == '2') |
    (col("id_carro") == '3') |
    (col("id_carro") == '4')
)

display(df_carros_source)

display(df_carros_final)
```

```python
# Subquery EXISTS: verifica existência de registros
df_carros_source.createOrReplaceTempView('source')
df_carros_final.createOrReplaceTempView('final')

df_carros_sql = spark.sql(
    """
    SELECT
        *
    FROM final as f
    WHERE EXISTS(
        SELECT * FROM source as s
        WHERE f.id_carro = s.id_carro
    )
    """
)
display(df_carros_sql)
```

```python
# EXISTS usando %sql magic
%sql
SELECT
    *
FROM final as f
WHERE EXISTS(
    SELECT * FROM source as s
    WHERE f.id_carro = s.id_carro
)
```

```python
# LEFT SEMI JOIN: equivalente ao EXISTS no PySpark
df_carros_spark = df_carros_final.join(
    df_carros_source,
    df_carros_final.id_carro == df_carros_source.id_carro,
    'left_semi'
)
display(df_carros_spark)
```
