#Aula 2 - PySpark

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, desc, asc
```

```python
sprk = SparkSession.builder.appName("Spark DataFrames").getOrCreate # cria uma sessão Spark
```

```python
spark = SparkSession.builder.getOrCreate()
data = [("Julio", 20), ("Raphael", 30), ("Steph", 26)]
cols = ["Nome", "Idade"]
df = spark.createDataFrame(data = data, schema = cols) # cria um dataframe usando uma lista de tuplas
df.display() # exibe o dataframe
```

#Operações com DataFrames

**select**

```python
df.select("Nome").display()
```

```python
df.select("Idade").display()
```

**filter**

```python
df.filter(df.Idade > 20).display()
```

```python
df.filter(df["Idade"] > 20).display()
```

```python
df.filter(col('Idade') > 20).display()
```

**order by**

```python
df.orderBy("Idade").display()
```

```python
df.orderBy("Nome").display()
```

```python
df.orderBy(asc("Idade")).display()
```

```python
df.orderBy(desc("Idade")).display()
```
