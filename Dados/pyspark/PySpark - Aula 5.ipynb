#Aula 5 - PySpark

```python
import pyspark
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, desc, asc, when, avg, sum
```

```python
data = [("Julio", "DE", 2000),
        ("May", "Marketing", 7000),
        ("Clara", "Eventos", 3000),
        ("Stephanie", "DE", 4700)]
columns = ["Nome", "Cargo", "Salario"]
df = spark.createDataFrame(data, schema=columns)
df.display()
```

```python
df_new = df.withColumn ("Categoria Salarial", when(col("Salario")>3000,"Descente").otherwise("Baixo"))
df_new.display()
```

**Agrupando e Agregando Dados**

```python
df_new.groupBy("Cargo").agg(avg("Salario").alias("Media")).display() # media salario por cargo
```

```python
df_new.groupBy("Cargo").agg(sum("Salario").alias("Total")).display() # somar salario por cargo 
```
