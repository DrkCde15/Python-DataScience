###Aula 8 - PySpark: Substring, right, left

```python
# Leitura do dataset de carros
#Imports
from pyspark.sql.functions import *
from pyspark.sql.types import *

#Leitura de arquivos
df_carros = spark.read.format('csv').option('header', 'true').option('encoding', 'utf-8').load('/Workspace/Users/jcesarsantana215@gmail.com/datasets/modelo_carro.csv', sep =',')

display(df_carros)
```

```python
# Funções de string: substring, left e right no SQL
df_carros.createOrReplaceTempView('carros')
df_carros_sql = spark.sql(
    """
    SELECT 
        modelo_carro,
        substring(modelo_carro, 2, 4),
        left(modelo_carro, 2),
        right(modelo_carro, 2)
    FROM carros
    """
)
display(df_carros_sql)
```

```python
# Extraindo substring com %sql magic
%sql
SELECT 
    modelo_carro,
    substring(modelo_carro, 2, 4)
FROM carros
```

```python
# Combinando substring e left
%sql
SELECT 
    modelo_carro,
    substring(modelo_carro, 2, 4),
    left(modelo_carro, 4)
FROM carros
```

```python
# Combinando substring e right
%sql
SELECT 
    modelo_carro,
    substring(modelo_carro, 2, 4),
    right(modelo_carro, 4)
FROM carros
```

```python
# Todas as funções de string juntas
%sql
SELECT 
    modelo_carro,
    substring(modelo_carro, 2, 4),
    left(modelo_carro, 2),
    right(modelo_carro, 2)
FROM carros
```

```python
# Funções de string usando PySpark API
df_carros_pyspark = df_carros

df_carros_pyspark = df_carros_pyspark.withColumn(
    "modelo_sub", substring("modelo_carro", 1, 2)
).withColumn(
    "modelo_right", expr("RIGHT(modelo_carro, 2)")
).withColumn(
    "modelo_left", expr("LEFT(modelo_carro, 2)")
)

display(df_carros_pyspark)
```
