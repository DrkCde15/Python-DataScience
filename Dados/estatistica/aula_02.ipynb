```python
import pandas as pd
```

```python
# Média
base = {
    'nums': [5, 7, 9, 100]
}
df = pd.DataFrame(base)

print(df.mean())
```

```python
base = {
    'nums': [1, 5, 100, 12, 28, 37, 9]
}

df = pd.DataFrame(base).sort_values(by='nums')

print(df.median())
```

```python
# Moda
base = {
    'nums': [5, 5, 7, 9, 100, 100, 100]
}
df = pd.DataFrame(base)

print(df.mode())
```
