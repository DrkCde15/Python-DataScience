# Exercício: Métricas de Classificação Binária

Avaliação de classificação binária com MLPClassifier no dataset make_moons.

**Conceitos:**
- Acurácia, Precisão, Recall, F1-Score
- Matriz de confusão
- `classification_report`
- Cálculo manual do F1-Score

# IMPORTS

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons, load_wine
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import (accuracy_score, classification_report,
                             confusion_matrix, precision_score, recall_score, f1_score)
```

# OPERAÇÕES

```python
x, y = make_moons(n_samples=200, noise=0.1, random_state=42)
```

```python
plt.scatter(x[:, 0], x[:, 1], c=y)
plt.grid()
plt.show()
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y)
model = MLPClassifier()
model.fit(x_train, y_train)
model.score(x_test, y_test)
```

```python
y_pred = model.predict(x_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
acuracia = accuracy_score(y_test, model.predict(x_test))
print("Acurácia:", acuracia)
```

```python
matriz_confusao = confusion_matrix(y_test, y_pred)
print(matriz_confusao)
```

```python
precisao = precision_score(y_test, y_pred)
print("Precisão:", precisao)
```

```python
recall = recall_score(y_test, y_pred)
print("Recall:", recall)
```

```python
f1 = f1_score(y_test, y_pred)
print("F1 Score:", f1)
```

```python
eq = 2 * precisao * recall / (precisao + recall)
print("F1 Score (manual):", eq)
```

```python
report = classification_report(y_test, y_pred)
print(report)
```
