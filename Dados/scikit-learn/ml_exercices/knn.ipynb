# IMPORTS

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from sklearn.datasets import make_classification, load_wine, load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor
from sklearn.metrics import confusion_matrix, mean_squared_error
```

# OPERAÇÕES

```python
x, y = make_classification(n_samples=200)
plt.scatter(x[:, 0], x[:, 1], c=y)
plt.grid()
plt.show()
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y)
plt.scatter(x_train[:, 0], x_train[:, 1], c=y_train)
plt.grid()
plt.show()
```

```python
plt.scatter(x_test[:, 0], x_test[:, 1], c=y_test)
plt.grid()
plt.show()
```

```python
plt.scatter(x_test[:, 0], x_test[:, 1], c=y_test)
plt.scatter(x_train[:, 0], x_train[:, 1], c=y_train)
plt.grid()
plt.show()
```

```python
model = KNeighborsClassifier()
model.fit(x_train, y_train)
model.score(x_test, y_test)
```

```python
y_pred = model.predict(x_test)
print(y_pred)
print(y_test)
```

```python
matrix = confusion_matrix(y_test, y_pred)
print(matrix)
```

```python
plt.scatter(x_test[2, 0], x_test[2, 1], s=120)
plt.scatter(x_train[:, 0], x_train[:, 1], c=y_train)
plt.grid()
plt.show()
```

```python
wine = load_wine()
print(wine.data.shape)
print(wine.target.shape)
print(wine.data)
```

```python
class_names = len(np.unique(wine.target))
print(class_names)
```

```python
x_train, x_test, y_train, y_test = train_test_split(wine.data, wine.target)

print("Training set shape:", x_train.shape)
print("Test set shape:", x_test.shape)
print("Training labels shape:", y_train.shape)
print("Test labels shape:", y_test.shape)
```

```python
model = KNeighborsClassifier()
model.fit(x_train, y_train)
model.score(x_test, y_test)
y_pred = model.predict(x_test)
print(y_pred)
```

```python
matriz_confusao = confusion_matrix(y_test, y_pred)
print(matriz_confusao)
```

# REGRESSÃO COM KNN

```python
data = load_diabetes()
print(data.data.shape)
print(data.target.shape)
print(data.feature_names)
print(data.data)
print(data.target)
```

```python
x_train, x_test, y_train, y_test = train_test_split(data.data, data.target)
model = KNeighborsRegressor()
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
print(y_pred)
```

```python
df_y_test = pd.DataFrame(y_test, columns=['y_test'])
df_y_pred = pd.DataFrame(y_pred, columns=['y_pred'])
pd.concat([df_y_test, df_y_pred], axis=1)
```

```python
plt.plot(y_test)
plt.plot(y_pred)
plt.grid()
plt.xlabel('x')
plt.ylabel('y')
plt.legend(['y_test', 'predicciones'])
plt.show()
```

```python
mean_squared_error(y_test, y_pred)
```
