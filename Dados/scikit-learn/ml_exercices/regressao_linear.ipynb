# IMPORT

```python
from sklearn.linear_model import LinearRegression
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# OPERAÇÕES

```python
# Leitura dos dados
df = pd.read_excel('./datasets/ReducaoOxigenio.xlsx')
x = df[['Reducao de solidos']]
y = df[['Reducao do oxigenio']]
```

```python
# Graficar os dados
plt.scatter(x, y)
plt.xlabel('Redução de sólidos')
plt.ylabel('Redução do oxigênio')
plt.grid()
plt.show()
```

```python
# Cálculo dos coeficientes da regressão linear

mtz = df.to_numpy()
n = len(mtz)
sum_x = np.sum(mtz[:, 0]) # Soma de x = soma de redução de sólidos
sum_y = np.sum(mtz[:, 1])# Soma de y = soma de redução do oxigênio
sum_xy = np.sum(mtz[:, 0] * mtz[:, 1]) # Soma do produto de x e y = soma do produto de redução de sólidos e redução do oxigênio
sum_x2 = np.sum([mtz[:, 0] * mtz[:, 0]]) # Soma de x ao quadrado = soma do quadrado da redução de sólidos

print(f'Quantidade de dados: {n}')
print(f'Soma de x: {sum_x}')
print(f'Soma de y: {sum_y}')
print(f'Soma do produto de x e y: {sum_xy}')
print(f'Soma de x ao quadrado: {sum_x2}')
```

```python
# Cálculo dos coeficientes da regressão linear

b1 = (n * sum_xy - sum_x * sum_y) / (n * sum_x2 - sum_x ** 2) # Coeficiente angular (b1) = (n * soma do produto de x e y - soma de x * soma de y) / (n * soma de x ao quadrado - soma de x ao quadrado)
b0 = (sum_y - b1 * sum_x) / n # Coeficiente linear (b0) = (soma de y - b1 * soma de x) / n

print(f'Coeficiente angular (b1): {b1}')
print(f'Coeficiente linear (b0): {b0}')
```

```python
# Cálculo dos coeficientes da regressão linear utilizando o scikit-learn

model = LinearRegression()
model.fit(x, y)
model.coef_ # Coeficiente angular (b1)
model.intercept_ # Coeficiente linear (b0)

print(int(model.predict([[10]]))) # Previsão para redução de sólidos igual a 10
```

```python
# Graficar a linha de regressão

plt.scatter(x, y)
plt.plot(x, model.predict(x), color='red') # Linha de regressão
plt.title('Regressão Linear')
plt.xlabel('Redução de sólidos')
plt.ylabel('Redução do oxigênio')
plt.legend(['y', 'Regressão Linear'])
plt.grid()
plt.show()
```
